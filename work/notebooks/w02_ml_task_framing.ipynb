## Why ML Beats a Fixed Rule
A fixed rule may use one or two conditions, such as low clicks or high impressions. However, content opportunity depends on multiple signals together, including clicks, impressions, CTR, search position, and search volume. ML can learn patterns from these signals and produce a ranking score for each page. This can provide more useful decision-support than relying on one fixed rule.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Compare the information used by a fixed rule and an ML approach

signals = [
    "clicks",
    "impressions",
    "CTR",
    "search position",
    "search volume"
]

fixed_rule_signals = 1
ml_signals = len(signals)

print("Fixed rule may use:", fixed_rule_signals, "or a few signals")
print("ML approach can use:", ml_signals, "signals")
print("Signals:", signals)

assert ml_signals > fixed_rule_signals

print("Why ML beats a fixed rule check: PASS")
```
